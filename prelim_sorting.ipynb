```python
def show_value_labels(variable, meta):
    """
    Display the value labels associated with a variable.
    """
    label_set_name = meta.variable_to_label.get(variable)

    print("Variable:", variable)
    print("Column label:", meta.column_names_to_labels.get(variable))
    print("Label-set name:", label_set_name)

    if label_set_name is None:
        print("No value labels found.")
    else:
        print(meta.value_labels.get(label_set_name, {}))
```

## Dimensionality Reduction

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA, SparsePCA
from kneed import KneeLocator

import prince
```

```python
NUMERIC_VARIANCE_THRESHOLD = 0.80
CATEGORICAL_INERTIA_THRESHOLD = 0.80

# Controls sparsity in Sparse PCA.
# Larger values produce more zero loadings.
SPCA_ALPHA = 1.0

RANDOM_STATE = 42
```

```python
constant_numeric_columns = [
    col
    for col in numeric_columns
    if cluster_features_complete_df[col].nunique(dropna=True) < 2
]

constant_categorical_columns = [
    col
    for col in categorical_columns
    if cluster_features_complete_df[col].nunique(dropna=True) < 2
]

constant_columns = (
    constant_numeric_columns
    + constant_categorical_columns
)

print("Constant numeric variables:")
print(constant_numeric_columns)

print("\nConstant categorical variables:")
print(constant_categorical_columns)
```

```python
cluster_features_dimension_df = (
    cluster_features_complete_df
    .drop(columns=constant_columns, errors="ignore")
    .copy()
)

numeric_columns = [
    col
    for col in numeric_columns
    if col not in constant_numeric_columns
]

categorical_columns = [
    col
    for col in categorical_columns
    if col not in constant_categorical_columns
]

print("\nChildren:", cluster_features_dimension_df.shape[0])
print("Numeric variables:", len(numeric_columns))
print("Categorical variables:", len(categorical_columns))
print(
    "Remaining missing values:",
    cluster_features_dimension_df.isna().sum().sum()
)
```

**Numeric Dimensionality Reduction**

```python
numeric_df = (
    cluster_features_dimension_df[numeric_columns]
    .astype(float)
    .copy()
)

numeric_scaler = StandardScaler()

numeric_standardized_array = (
    numeric_scaler.fit_transform(numeric_df)
)

numeric_standardized_df = pd.DataFrame(
    numeric_standardized_array,
    columns=numeric_columns,
    index=numeric_df.index
)

numeric_standardized_df.head()
```

```python
pd.DataFrame({
    "mean": numeric_standardized_df.mean(),
    "standard_deviation": numeric_standardized_df.std(ddof=0)
})
```

```python
maximum_numeric_components = min(
    numeric_standardized_df.shape[0],
    numeric_standardized_df.shape[1]
)

pca_selection_model = PCA(
    n_components=maximum_numeric_components
)

pca_selection_scores = pca_selection_model.fit_transform(
    numeric_standardized_df
)

pca_explained_variance_ratio = (
    pca_selection_model.explained_variance_ratio_
)

pca_cumulative_variance = np.cumsum(
    pca_explained_variance_ratio
)
```

```python
pca_selection_summary_df = pd.DataFrame({
    "component": np.arange(
        1,
        len(pca_explained_variance_ratio) + 1
    ),
    "explained_variance_ratio": (
        pca_explained_variance_ratio
    ),
    "cumulative_explained_variance": (
        pca_cumulative_variance
    )
})

pca_selection_summary_df
```

```python
pca_knee = KneeLocator(
    x=pca_selection_summary_df["component"],
    y=pca_selection_summary_df["explained_variance_ratio"],
    curve="convex",
    direction="decreasing"
)

pca_elbow_components = pca_knee.knee

if pca_elbow_components is None:
    pca_elbow_components = 1

print(
    "Number of components from PCA elbow:",
    pca_elbow_components
)
```

```python
pca_variance_components = (
    np.argmax(
        pca_cumulative_variance
        >= NUMERIC_VARIANCE_THRESHOLD
    )
    + 1
)

print(
    f"Components required to explain at least "
    f"{NUMERIC_VARIANCE_THRESHOLD:.0%} of variance:",
    pca_variance_components
)
```

```python
n_numeric_components = max(
    int(pca_elbow_components),
    int(pca_variance_components)
)

n_numeric_components = min(
    n_numeric_components,
    maximum_numeric_components
)

print(
    "Final number of numeric components:",
    n_numeric_components
)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    pca_selection_summary_df["component"],
    pca_selection_summary_df["explained_variance_ratio"],
    marker="o"
)

plt.axvline(
    pca_elbow_components,
    linestyle="--",
    label=f"Elbow = {pca_elbow_components}"
)

plt.xlabel("Principal component")
plt.ylabel("Explained variance ratio")
plt.title("PCA scree plot for numeric variables")
plt.xticks(
    pca_selection_summary_df["component"]
)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    pca_selection_summary_df["component"],
    pca_selection_summary_df[
        "cumulative_explained_variance"
    ],
    marker="o"
)

plt.axhline(
    NUMERIC_VARIANCE_THRESHOLD,
    linestyle="--",
    label=(
        f"Threshold = "
        f"{NUMERIC_VARIANCE_THRESHOLD:.0%}"
    )
)

plt.axvline(
    n_numeric_components,
    linestyle="--",
    label=(
        f"Selected components = "
        f"{n_numeric_components}"
    )
)

plt.xlabel("Number of components")
plt.ylabel("Cumulative explained variance")
plt.title("PCA cumulative explained variance")
plt.ylim(0, 1.05)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
spca_model = SparsePCA(
    n_components=n_numeric_components,
    alpha=SPCA_ALPHA,
    random_state=RANDOM_STATE,
    max_iter=5000,
    tol=1e-6
)

spca_score_array = spca_model.fit_transform(
    numeric_standardized_df
)
```

```python
spca_component_names = [
    f"SPCA_{i}"
    for i in range(1, n_numeric_components + 1)
]

spca_scores_df = pd.DataFrame(
    spca_score_array,
    columns=spca_component_names,
    index=numeric_standardized_df.index
)

spca_scores_df.head()
```

```python
spca_loadings_df = pd.DataFrame(
    spca_model.components_.T,
    index=numeric_columns,
    columns=spca_component_names
)

spca_loadings_df
```

```python
spca_correlations_df = pd.DataFrame(
    index=numeric_columns,
    columns=spca_component_names,
    dtype=float
)

for variable in numeric_columns:
    for component in spca_component_names:
        spca_correlations_df.loc[
            variable,
            component
        ] = numeric_standardized_df[
            variable
        ].corr(
            spca_scores_df[component]
        )

spca_correlations_df
```

```python
spca_correlation_long_df = (
    spca_correlations_df
    .reset_index(names="variable")
    .melt(
        id_vars="variable",
        var_name="component",
        value_name="pearson_correlation"
    )
)

spca_correlation_long_df[
    "absolute_correlation"
] = (
    spca_correlation_long_df[
        "pearson_correlation"
    ].abs()
)

spca_correlation_long_df = (
    spca_correlation_long_df
    .sort_values(
        ["component", "absolute_correlation"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

spca_correlation_long_df
```

```python
print("MCA scores shape:", mca_scores_df.shape)
print(
    "Missing MCA scores:",
    mca_scores_df.isna().sum().sum()
)
```

```python
def partial_eta_squared(
    component_scores,
    categorical_variable
):
    analysis_data = pd.DataFrame({
        "score": pd.to_numeric(
            component_scores,
            errors="coerce"
        ),
        "group": categorical_variable
    }).dropna()

    if analysis_data.empty:
        return np.nan

    if analysis_data["group"].nunique() < 2:
        return np.nan

    grand_mean = analysis_data["score"].mean()

    ss_between = 0.0
    ss_within = 0.0

    for _, group_df in analysis_data.groupby(
        "group"
    ):
        group_scores = group_df["score"]
        group_mean = group_scores.mean()

        ss_between += (
            len(group_scores)
            * (group_mean - grand_mean) ** 2
        )

        ss_within += (
            (group_scores - group_mean) ** 2
        ).sum()

    denominator = ss_between + ss_within

    if denominator == 0:
        return 0.0

    return ss_between / denominator
```

```python
mca_eta_squared_df = pd.DataFrame(
    index=categorical_columns,
    columns=mca_component_names,
    dtype=float
)

for variable in categorical_columns:
    for component in mca_component_names:

        mca_eta_squared_df.loc[
            variable,
            component
        ] = partial_eta_squared(
            mca_scores_df[component],
            categorical_df[variable]
        )

mca_eta_squared_df
```

```python
mca_eta_squared_long_df = (
    mca_eta_squared_df
    .reset_index(names="variable")
    .melt(
        id_vars="variable",
        var_name="component",
        value_name="partial_eta_squared"
    )
    .sort_values(
        ["component", "partial_eta_squared"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

mca_eta_squared_long_df
```
